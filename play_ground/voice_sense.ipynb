```python
from pydub import AudioSegment
import webrtcvad

def read_frames(audio, frame_duration_ms=30):
    """将音频分割成帧"""
    n = int(frame_duration_ms * audio.frame_rate / 1000)
    return [audio[i:i+n] for i in range(0, len(audio), n)]

def is_speech(frame, vad, sample_rate):
    """判断帧是否包含语音"""
    byte_frame = frame.raw_data
    return vad.is_speech(byte_frame, sample_rate)

def remove_silence(audio, vad, frame_duration_ms=30):
    """去除音频中的空白部分"""
    frames = read_frames(audio, frame_duration_ms)
    speech_frames = [frame for frame in frames if is_speech(frame, vad, audio.frame_rate)]
    return sum(speech_frames)

# 加载WAV文件
audio = AudioSegment.from_wav("../wav/tesla_2024_q3.wav")

# 初始化VAD
vad = webrtcvad.Vad()
vad.set_mode(3)  # 设置VAD模式，0-3，数字越大越严格

# 去除空白部分
cleaned_audio = remove_silence(audio, vad)

# 保存处理后的音频
cleaned_audio.export("output.wav", format="wav")
```
